```python
# Importing the dependencies
```

```python
import numpy as np
import pandas as pd
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

```python
# loading the diabetes dataset to a pandas DataFrame
```

```python
diabetes_dataset = pd.read_csv("diabetes.csv")
```

```python
diabetes_dataset.head()
```

```python
# number of rows and columns in the dataset
```

```python
diabetes_dataset.shape
```

```python
#getting the statistical measures of the data
```

```python
diabetes_dataset.describe()
```

```python
diabetes_dataset['Outcome'].value_counts()
```

```python
# 0--> Non-Diabetic , 1 --> Diabetic
```

```python
diabetes_dataset.groupby('Outcome').mean()
```

```python
# separating the features and labels
```

```python
x = diabetes_dataset.drop(columns = 'Outcome', axis = 1)  #features
```

```python
y = diabetes_dataset['Outcome']           #labels
```

```python
print(x)
```

```python
print(y)
```

```python
# Data Standardization
```

```python
scaler = StandardScaler()
```

```python
standardized_data = scaler.fit_transform(x)
```

```python
print(standardized_data)
```

```python
x = standardized_data
y = diabetes_dataset['Outcome']
```

```python
print(x)
print(y)
```

```python
# Train - Test Split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, stratify=y, random_state=2)
```

```python
print(x.shape, x_train.shape, x_test.shape)
```

```python
# Training the data
```

```python
classifier = svm.SVC(kernel='linear')
```

```python
#training the support vector machine classifier
classifier.fit(x_train, y_train)
```

```python
# Accuracy Score on the Training data
```

```python
x_train_prediction = classifier.predict(x_train)
training_data_accuracy = accuracy_score(x_train_prediction, y_train)
print('Accuracy score of the training data: ',training_data_accuracy)
```

```python
# Accuracy Score on the Testing data
```

```python
x_test_prediction = classifier.predict(x_test)
testing_data_accuracy = accuracy_score(x_test_prediction, y_test)
print('Accuracy score of the testing data: ',testing_data_accuracy)  
```

```python
# Making the Predictive System
```

```python
input_data = (10,168,74,0,0,38,0.537,34)

# changing the input data to numpy array
input_data_as_numpy_array = np.asarray(input_data)
print("Numpy Array: ",input_data_as_numpy_array)

# reshape the function as we are predicting for one instance
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)
print("Reshaped Numpy Array: ",input_data_reshaped )

# standardize the input data
# std_data = scaler.transform(input_data_reshaped)
# print("Standardized Numpy Array: ",std_data)

# Predicting whether particular person is diabetic or not
prediction = classifier.predict(input_data_reshaped)
print(prediction)

if(prediction[0]==0):
    print("The person is not Diabetic")
else:
    print("The person is Diabetic")
```
